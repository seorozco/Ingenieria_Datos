# 02 - Lectura de Validación

**Objetivo:** Leer y analizar resultados de validación de datos

**Entrada:**
- Tabla: `raw.validacion_calidad`

**Salida:**
- DataFrame con validaciones
- Archivo: `datos/bronze/validacion_export.csv`

```python
import pandas as pd
from pathlib import Path
import sys

sys.path.insert(0, str(Path.cwd().parent))
from conexion_sqlserver import obtener_engines

print("Librerías importadas correctamente")
```

## 1. Conectar a BD

```python
engine_origen, engine_destino = obtener_engines()

print("Conectado a SQL Server")
```

## 2. Leer datos de validación

```python
# Leer tabla de validación
query = "SELECT * FROM raw.validacion_calidad"
df_validacion = pd.read_sql_query(query, engine_destino)

print(f"Registros de validación: {len(df_validacion)}")
print(f"Columnas: {list(df_validacion.columns)}")
```

## 3. Analizar resultados

```python
# Contar por tipo de error
conteos = df_validacion['tipo_error'].value_counts()

print("\nResultados de Validacion:")
for error_type, count in conteos.items():
    print(f"  {error_type}: {count} registros")

# Mostrar primeros registros
print(f"\nPrimeros registros:")
print(df_validacion[['numero_fila', 'tipo_error', 'campo_con_error']].head(10))
```

## 4. Guardar en CSV

```python
datos_dir = Path.cwd() / "datos" / "bronze"
datos_dir.mkdir(parents=True, exist_ok=True)

csv_path = datos_dir / "validacion_export.csv"
df_validacion.to_csv(csv_path, index=False, encoding='utf-8')

print(f"Archivo guardado: {csv_path}")
print(f"Tamaño: {csv_path.stat().st_size / 1024:.1f} KB")
```

## Resumen

```python
ok_count = len(df_validacion[df_validacion['tipo_error'] == 'OK'])
error_count = len(df_validacion[df_validacion['tipo_error'] != 'OK'])
calidad = (ok_count / len(df_validacion)) * 100

print("\n" + "="*60)
print("RESUMEN - ETAPA 02: VALIDACION")
print("="*60)
print(f"Total validaciones: {len(df_validacion)}")
print(f"Estado OK: {ok_count}")
print(f"Con errores: {error_count}")
print(f"Tasa de calidad: {calidad:.1f}%")
print(f"Tabla origen: raw.validacion_calidad")
print(f"Archivo CSV: {csv_path.name}")
print("="*60)
```
